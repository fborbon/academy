## 1. El corpus

Nueve discursos, uno por época: Truman (1950), Eisenhower (1957), Kennedy (1962), Johnson (1966), Nixon (1974),
Reagan (1982), Clinton (1994) y G. W. Bush (2002 y 2006). Proceden del corpus `state_union` que distribuye NLTK, que
reproduce los textos del American Presidency Project; al ser obras del gobierno federal, son de dominio público.

Dividimos cada discurso en **fragmentos** de unas 50 palabras formados por frases completas, y ponemos una etiqueta delante de
cada fragmento (`1974 address by President Nixon.`). Si la etiqueta ayuda se mide en la sección 3, y la respuesta es menos
clara de lo que nos gustaría.

```python
import os, re, math, glob, urllib.request
from collections import Counter, defaultdict
import numpy as np
import matplotlib.pyplot as plt
from sentence_transformers import SentenceTransformer

SPEECHES = ["1950-Truman", "1957-Eisenhower", "1962-Kennedy", "1966-Johnson", "1974-Nixon",
            "1982-Reagan", "1994-Clinton", "2002-GWBush", "2006-GWBush"]
DATA = next((p for p in ("data/sotu", "../data/sotu") if os.path.isdir(p)), "data/sotu")
os.makedirs(DATA, exist_ok=True)
for name in SPEECHES:                       # solo hace falta si descargaste este notebook por separado
    path = f"{DATA}/{name}.txt"
    if not os.path.exists(path):
        urllib.request.urlretrieve(f"https://raw.githubusercontent.com/fborbon/academy/main/rag-systems/data/sotu/{name}.txt", path)

def read_sentences(name):
    paras = [p.strip() for p in open(f"{DATA}/{name}.txt", encoding="utf-8").read().split("\n")
             if len(p.strip()) > 60 and not p.strip().startswith("Note:") and not p.strip().isupper()]
    return [s for p in paras for s in re.split(r"(?<=[.!?])\s+", p) if s]

def make_chunks(target_words):
    # consecutive whole sentences until the chunk reaches target_words; the speech, the year and the speaker travel with it
    out = []
    for name in SPEECHES:
        year, pres = name.split("-")
        cur, n, k = [], 0, 1
        for s in read_sentences(name):
            cur.append(s); n += len(s.split())
            if n >= target_words:
                out.append(dict(id=f"{year}-{k:03d}", year=year, pres=pres, text=" ".join(cur))); cur, n, k = [], 0, k + 1
        if cur:
            out.append(dict(id=f"{year}-{k:03d}", year=year, pres=pres, text=" ".join(cur)))
    return out

chunks = make_chunks(50)
print(f"{len(SPEECHES)} speeches, {sum(len(read_sentences(n)) for n in SPEECHES)} sentences, {len(chunks)} chunks of about 50 words")
print(f"{'speech':<18}{'words':>8}{'chunks':>8}")
for n in SPEECHES:
    print(f"{n:<18}{sum(len(s.split()) for s in read_sentences(n)):>8}{sum(c['year'] == n[:4] for c in chunks):>8}")
```

## 3. Recuperación: qué medir antes de construir nada

Un recuperador es bueno para una pregunta si algún fragmento con la clave llega a lo más alto del ranking. Tres decisiones,
cada una **elegida solo con el conjunto de desarrollo**: qué tamaño tiene un fragmento, si el orador y el año se escriben dentro del fragmento, y
cuánto BM25 se mezcla con los embeddings (módulo 02).

```python
embedder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")
STOP = set("a an and are as at be by can do does for from how i if in is it of on or that the this to what when which who with you my me did was were has have had been will".split())
tokens = lambda text: [w for w in re.findall(r"[a-z0-9]+", text.lower()) if w not in STOP]

class Index:
    # BM25 + embeddings over a list of chunks; label=True writes the year and speaker in front of each chunk
    def __init__(self, chunks, label=True):
        self.chunks = chunks
        self.texts = [(f"{c['year']} address by President {c['pres']}. " if label else "") + c["text"] for c in chunks]
        self.tf = [Counter(tokens(t)) for t in self.texts]
        self.len = np.array([sum(d.values()) for d in self.tf], float)
        df = Counter(w for d in self.tf for w in d)
        n = len(self.texts)
        self.idf = {w: math.log(1 + (n - c + 0.5) / (c + 0.5)) for w, c in df.items()}
        self.emb = embedder.encode(self.texts, normalize_embeddings=True, batch_size=64)
        self.avg = self.len.mean()
    def bm25(self, q):
        s = np.zeros(len(self.texts))
        for w in set(tokens(q)):
            if w in self.idf:
                for i, d in enumerate(self.tf):
                    f = d.get(w, 0)
                    if f:
                        s[i] += self.idf[w] * f * 2.5 / (f + 1.5 * (0.25 + 0.75 * self.len[i] / self.avg))
        return s
    def cosine(self, q):
        return self.emb @ embedder.encode(q, normalize_embeddings=True)
    def fused(self, q, w_bm25):
        tot = np.zeros(len(self.texts))
        for sc, w in ((self.bm25(q), w_bm25), (self.cosine(q), 1.0)):
            for rank, i in enumerate(np.argsort(-sc), 1):
                tot[i] += w / (60 + rank)
        return tot

def retrieval(idx, S, w):
    # hit@1, hit@3, MRR for the answerable questions of S
    ranks = []
    for _, q, year, key, _t in S:
        if year is None:
            continue
        order = list(np.argsort(-idx.fused(q, w)))
        gold = [i for i, c in enumerate(idx.chunks) if c["year"] == year and key in c["text"]]
        ranks.append(min(order.index(g) for g in gold) + 1)
    r = np.array(ranks)
    return (r == 1).mean(), (r <= 3).mean(), (1 / r).mean()

print(f"{'chunk size':<14}{'chunks':>7}   dev hit@1  hit@3   MRR      battery hit@1  hit@3   MRR")
SIZES = {}
for target in (25, 50, 100, 200):
    ch = make_chunks(target); idx_t = Index(ch)
    SIZES[target] = (idx_t, retrieval(idx_t, DEV, 0.5), retrieval(idx_t, BATTERY, 0.5))
    d, b = SIZES[target][1], SIZES[target][2]
    print(f"{target:>3} words{len(ch):>10}      {d[0]:>6.2f} {d[1]:>6.2f} {d[2]:>6.2f}          {b[0]:>6.2f} {b[1]:>6.2f} {b[2]:>6.2f}")
CHUNK_WORDS = max(SIZES, key=lambda t: (SIZES[t][1][2], -t))
print("chosen on the dev set (best MRR):", CHUNK_WORDS, "words")
```

El **conjunto de desarrollo elige un tamaño de fragmento y la batería habría elegido otro**. Con 14 y 22 preguntas con respuesta
respectivamente, una diferencia de una o dos preguntas cambia el ranking: la misma advertencia del módulo 07, ahora sobre texto real.
Mantenemos la elección hecha en desarrollo, porque elegir con la batería sería filtrar información. Ambas medidas coinciden en que
los fragmentos muy grandes (200 palabras, varios temas mezclados en un vector) son una mala idea para estos discursos.

A continuación, dos elecciones más, de nuevo en desarrollo: si la etiqueta (orador y año) va delante de cada fragmento, y el peso de BM25.

```python
P = SIZES[CHUNK_WORDS][0]
labelled = P
plain = Index(P.chunks, label=False)
WHO = [q for q in DEV + BATTERY if q[4] == "who"]
print("speaker and year written into the chunk? (BM25 weight 0.5, chunk size", CHUNK_WORDS, ")")
print(f"{'':<18}{'who questions hit@1':>22}{'all dev hit@1':>16}{'all battery hit@1':>20}")
for name, ix in (("no label", plain), ("with label", labelled)):
    print(f"{name:<18}{retrieval(ix, WHO, 0.5)[0]:>22.2f}{retrieval(ix, DEV, 0.5)[0]:>16.2f}{retrieval(ix, BATTERY, 0.5)[0]:>20.2f}")

print("\nBM25 weight (semantic weight = 1) on the DEV set:")
dev_w = {}
for w in (0, 0.25, 0.5, 1, 2):
    dev_w[w] = retrieval(labelled, DEV, w)
    print(f"  weight {w:<5} hit@1 {dev_w[w][0]:.2f}  hit@3 {dev_w[w][1]:.2f}  MRR {dev_w[w][2]:.2f}")
W_BM25 = max(dev_w, key=lambda w: (dev_w[w][2], -w))
print("chosen BM25 weight:", W_BM25)
```

Tres cosas que leer en esta tabla, ninguna de las que te diría un tutorial:

* **La etiqueta ayuda en desarrollo y perjudica en la batería.** En las preguntas "who" no cambia nada. Ambos conjuntos son diminutos, así que no podemos decir que ayude; la mantenemos, como se eligió en desarrollo, y **no** afirmamos ningún beneficio.
* **La búsqueda híbrida sí ayuda aquí**, con un peso pequeño de BM25 (0,25). En el proyecto final el mejor peso era cero. El peso correcto depende del corpus: estos discursos están llenos de cifras y nombres que BM25 encuentra exactamente. Es la lección del módulo 02 otra vez, con el signo contrario.
* **El peso hay que medirlo, no recordarlo.** Una regla "lo híbrido siempre es mejor" y una regla "lo híbrido nunca es mejor" son ambas falsas.

## 4. El primer sistema: el método del proyecto final, sin cambios

Mismo diseño que el proyecto final: recuperar, rechazar cuando el mejor coseno está por debajo de un umbral, y responder con la frase del
fragmento principal más cercana a la pregunta. El umbral sale de las preguntas con respuesta de desarrollo y de las preguntas fuera de tema.

```python
SENT = {}                                                    # sentence embeddings cached per chunk
def sentences_of(i):
    if i not in SENT:
        s = [x for x in re.split(r"(?<=[.!?])\s+", P.chunks[i]["text"]) if x]
        SENT[i] = (s, embedder.encode(s, normalize_embeddings=True))
    return SENT[i]

dev_top = [P.cosine(q[1]).max() for q in DEV if q[2]]
off_top = [P.cosine(q).max() for q in OFF_TOPIC]
T_COS = (max(off_top) + min(dev_top)) / 2
print(f"top cosine: answerable dev {min(dev_top):.2f} to {max(dev_top):.2f}; off-topic {min(off_top):.2f} to {max(off_top):.2f}  ->  refuse below {T_COS:.2f}")

def answer_v1(q):
    cos = P.cosine(q)
    if cos.max() < T_COS:
        return "refuse", None, None
    top = int(np.argmax(P.fused(q, W_BM25)))
    s, e = sentences_of(top)
    return "answer", P.chunks[top]["id"], s[int(np.argmax(e @ embedder.encode(q, normalize_embeddings=True)))]

def grade(q, res):
    kind, cid, sent = res
    if q[2] is None:
        return "ok" if kind == "refuse" else "wrong"
    if kind == "refuse":
        return "missed"
    return "ok" if cid[:4] == q[2] and q[3] in sent else "wrong"

def report(system, S):
    res = [(q, grade(q, system(q[1]))) for q in S]
    by = defaultdict(lambda: [0, 0])
    for q, g in res:
        by[q[4]][1] += 1; by[q[4]][0] += g == "ok"
    return sum(g == "ok" for _, g in res), len(S), dict(by), res

for name, S in (("dev", DEV), ("battery", BATTERY)):
    ok, n, by, res = report(answer_v1, S)
    print(f"{name:<8} correct {ok:>2} of {n}   " + "  ".join(f"{t} {a}/{b}" for t, (a, b) in by.items()))
```

El primer sistema acierta más o menos la mitad de la batería, muy por debajo del 75 % del proyecto final. Falla de dos maneras distintas, y
necesitan arreglos distintos:

* **Las preguntas sin respuesta nunca se rechazan.** El tema de "what did Reagan say about the Challenger disaster" está cerca de muchas frases
  del corpus, así que el coseno máximo es tan alto como en una pregunta real. Un umbral de similitud no puede separarlas.
* **Discurso correcto con frase equivocada, o discurso equivocado.** Varias preguntas devuelven un fragmento de otro año, o el fragmento correcto con una frase que solo
  repite las palabras de la pregunta ("How many men who listen to me tonight...").

```python
ok, n, by, res = report(answer_v1, BATTERY)
print("battery failures of the first system:")
for q, g in res:
    if g != "ok":
        kind, cid, sent = answer_v1(q[1])
        print(f"  {q[0]} [{g}] {q[1][:62]}\n        -> {cid} {(sent or '')[:78]}")
```

## 5. Tres arreglos que viven del lado de la consulta

Un supervisor que leyera estos fallos (módulo 08) los marcaría y pediría a una persona que arreglara la documentación. Aquí la
documentación es un registro histórico, así que los arreglos van al sistema que la rodea. Cada uno es una regla **general**, no un parche para
una pregunta, y cada uno tiene un parámetro que elegimos en el **conjunto de desarrollo**:

1. **Filtro de orador y año.** Si la pregunta nombra a un presidente o un año, se busca solo en los discursos correspondientes. Un año sin discurso (2010) significa que no hay nada que encontrar.
2. **Comprobación de cobertura para rechazar.** En una pregunta con respuesta, casi todas sus palabras de contenido (con un stemming ligero) aparecen en los mejores fragmentos. Con "Challenger", "Katrina" o "Internet" no.
3. **Refuerzo de cantidades.** En preguntas "how many", "how much" o "what share", se prefiere la frase que contiene un número.

```python
PRES = {"truman": {"1950"}, "eisenhower": {"1957"}, "kennedy": {"1962"}, "johnson": {"1966"}, "nixon": {"1974"},
        "reagan": {"1982"}, "clinton": {"1994"}, "bush": {"2002", "2006"}}
ALL_YEARS = set(c["year"] for c in P.chunks)
GENERIC = set("president said say address speech tell tonight states united american americans year years much many".split())
QTY = re.compile(r"\b(how many|how much|what share|how far|how fast|how long|what increase)\b", re.I)

def stem(w):
    for suf in ("ing", "ed", "es", "s"):
        if w.endswith(suf) and len(w) - len(suf) >= 4:
            return w[:-len(suf)]
    return w

def allowed_speeches(q):
    named = [y for n, ys in PRES.items() if n in q.lower() for y in ys]
    years = set(re.findall(r"\b(19[4-9]\d|20[01]\d)\b", q))
    allowed = set(named) if named else set(ALL_YEARS)
    return allowed & years if years else allowed                # empty set: the question asks about a year we do not have

def coverage(q, chunk_ids):
    words = [stem(w) for w in set(tokens(q)) if w not in GENERIC and w not in PRES and not w.isdigit()]
    if not words:
        return 1.0
    seen = set(stem(w) for i in chunk_ids for w in tokens(P.chunks[i]["text"]))
    return sum(w in seen for w in words) / len(words)

def make_system(cov, boost, use_filter=True, k=3):
    def system(q):
        allowed = allowed_speeches(q) if use_filter else ALL_YEARS
        if not allowed:
            return "refuse", None, None
        sc = P.fused(q, W_BM25)
        for i, c in enumerate(P.chunks):
            if c["year"] not in allowed:
                sc[i] = -1
        order = list(np.argsort(-sc))[:k]
        if cov and coverage(q, order) < cov:
            return "refuse", None, None
        cand = [(i, s, e) for i in order for s, e in zip(*sentences_of(i))]
        sims = np.array([c[2] @ embedder.encode(q, normalize_embeddings=True) for c in cand])
        if boost and QTY.search(q):
            sims = sims + boost * np.array([bool(re.search(r"\d|\$|percent|million|billion", c[1])) for c in cand])
        i, s, _ = cand[int(np.argmax(sims))]
        return "answer", P.chunks[i]["id"], s
    return system

print(f"{'coverage':>9}{'boost':>7}   dev correct (of {len(DEV)})")
grid = {}
for cov in (0, 0.3, 0.4, 0.5, 0.6, 0.7):
    for boost in (0, 0.1, 0.2):
        grid[(cov, boost)] = report(make_system(cov, boost), DEV)[0]
        print(f"{cov:>9}{boost:>7}   {grid[(cov, boost)]:>3}")
COV, BOOST = max(grid, key=lambda k: (grid[k], k[0], -k[1]))
print("chosen on the dev set:", dict(coverage=COV, boost=BOOST))
```

## 6. Antes y después, en la batería y en el conjunto retenido

Las preguntas retenidas se escribieron después de diseñar los arreglos y nunca se usaron para elegir ningún parámetro. El
intervalo bootstrap (módulo 07) nos dice cuánto de cada cambio podría ser suerte con este tamaño.

```python
rng = np.random.default_rng(0)
def boot_diff(a, b, n=4000):
    # a, b: per-question 0/1 correctness for the same questions; the interval of mean(b) - mean(a)
    a, b = np.array(a, float), np.array(b, float)
    d = [(b[s] - a[s]).mean() for s in (rng.integers(0, len(a), len(a)) for _ in range(n))]
    return np.percentile(d, [2.5, 97.5])

v1, v2 = answer_v1, make_system(COV, BOOST)
rows = {}
print(f"{'set':<10}{'questions':>10}{'first system':>14}{'improved':>10}{'change':>9}   95% interval of the change")
for name, S in (("dev", DEV), ("battery", BATTERY), ("held-out", HELDOUT)):
    a = [g == "ok" for _, g in report(v1, S)[3]]
    b = [g == "ok" for _, g in report(v2, S)[3]]
    lo, hi = boot_diff(a, b)
    rows[name] = (np.mean(a), np.mean(b))
    print(f"{name:<10}{len(S):>10}{np.mean(a):>14.2f}{np.mean(b):>10.2f}{np.mean(b) - np.mean(a):>+9.2f}   [{lo:+.2f}, {hi:+.2f}]")

fig, ax = plt.subplots(figsize=(7, 3.2))
x = np.arange(3)
ax.bar(x - 0.18, [rows[k][0] for k in rows], 0.36, label="first system", color="#8a8aa0")
ax.bar(x + 0.18, [rows[k][1] for k in rows], 0.36, label="improved", color="#E1C340")
ax.set_xticks(x); ax.set_xticklabels(list(rows)); ax.set_ylim(0, 1); ax.set_ylabel("correct"); ax.legend(frameon=False)
ax.set_title("State of the Union assistant: share of correct answers")
plt.tight_layout(); plt.show()
```

```python
ok, n, by, res = report(v2, BATTERY)
print("improved system, battery:", ok, "of", n, " ", "  ".join(f"{t} {a}/{b}" for t, (a, b) in by.items()))
for q, g in res:
    if q[2] is None and g != "ok":
        print("  an unanswerable question that was still answered:", q[0], q[1], "->", v2(q[1])[1])
ok, n, by, res = report(v2, HELDOUT)
print("improved system, held-out:", ok, "of", n, " ", "  ".join(f"{t} {a}/{b}" for t, (a, b) in by.items()))
print("\nwhat still fails on the held-out set:")
for q, g in res:
    if g != "ok":
        kind, cid, sent = v2(q[1])
        print(f"  {q[0]} [{g}] {q[1][:60]}\n        -> {cid} {(sent or '')[:78]}")
```

## 7. Lo que enseñó este corpus y el proyecto final no podía

Resultados, en el orden de cuánto podemos fiarnos de ellos:

* **Rechazar lo que nunca se dijo es la mejora real.** Las preguntas sin respuesta pasaron de 0 de 6 en la batería a 5 de 6, y de 0 de 4 a 3 de 4 en preguntas retenidas que los arreglos nunca vieron. Esta mejora es grande y generalizó. Un umbral de similitud por sí solo no podía lograrlo, porque el tema de una pregunta sin respuesta suele estar cerca de algo que **sí** se dijo, en otro discurso o en una frase cercana.
* **La única pregunta sin respuesta que aún recibe una respuesta muestra una debilidad de la regla.** "What did the president say about the Internet?" solo tiene dos palabras de contenido tras quitar las genéricas, y la palabra de relleno "about" aparece en todos los fragmentos, así que la cobertura es del 50 %, justo en el umbral. Una lista de palabras genéricas crece con cada fallo; ponderar las palabras por IDF sería el arreglo con fundamento (ejercicio 3).
* **La mejora global es plausible pero no está demostrada.** Las respuestas correctas pasaron de 0,50 a 0,68 en la batería y de 0,44 a 0,62 en las preguntas retenidas, pero todos los intervalos del 95 % incluyen el cero. Con 28 y 16 preguntas es lo mejor que podemos decir, y el módulo 07 ya calculó cuántas preguntas harían falta para estar seguros.
* **El texto real es mucho más difícil que un texto escrito por ti.** El primer sistema sacó 0,75 en el corpus del proyecto final y 0,50 aquí con el mismo método. Documentos largos y retóricos, un recuperador que mezcla épocas y respuestas enterradas en frases que repiten la pregunta cuestan precisión. Una buena nota en tu propio corpus de prueba no es evidencia sobre documentos reales.
* **Los arreglos cuestan cobertura.** La comprobación de cobertura rechaza algunas preguntas con respuesta cuya redacción está lejos del texto ("How many private peace talks took place in 1965?"). Toda regla de rechazo cambia respuestas erróneas por respuestas perdidas, y esta se ajustó con 14 preguntas con respuesta.

**Qué cambia en el bucle de mejora.** El proyecto final reparó la documentación. Aquí el supervisor puede seguir marcando el fallo y una persona puede seguir leyendo los casos marcados, pero el arreglo va al lado de la consulta: un filtro, una regla de rechazo, una preferencia de ranking. Esos arreglos son generales, por eso se transfirieron a preguntas nuevas; un parche escrito para una pregunta que fallaba no lo habría hecho.
